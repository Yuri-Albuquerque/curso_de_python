## 2. JupyterLab em 5 minutos
O JupyterLab é a interface; o **kernel** é o processo Python que executa o código.
"Reiniciar o kernel" (atalho `0-0` duas vezes, ou menu *Kernel*) apaga tudo que
está na memória — é o jeito limpo de "ligar o carro de novo" quando o estado da
sessão vira uma bagunça.

**Modos de uma célula** (o contorno azul/da esquerda indica o modo):

- **Edição** (borda verde): cursor dentro da célula — você digita código/texto.
  Entre no modo com `Enter`.
- **Comando** (borda azul): célula selecionada sem cursor — atalhos funcionam.
  Saia do modo com `Esc`.

**Atalhos essenciais** (decorar `Shift+Enter` hoje; o resto vem com o uso):

| Atalho | Modo | Ação |
|---|---|---|
| `Shift+Enter` | ambos | executa a célula e vai para a próxima |
| `A` / `B` | Comando | nova célula **A**cima / **B**aixo |
| `DD` | Comando | apaga a célula |
| `M` / `Y` | Comando | converte para **M**arkdown / código |
| `0-0` | Comando | reinicia o kernel |

## 3. Ajudas e mágicos essenciais do IPython
No IPython/Jupyter, tudo é objeto — e todo objeto tem documentação a um `?` de
distância. Além disso, comandos prefixados por `%` são **comandos mágicos**
(magics): atalhos que só existem no ambiente interativo (VanderPlas, Cap. 1,
"IPython Magic Commands").

- `objeto?` → docstring; `objeto??` → código-fonte (quando disponível);
- `%who`, `%whos` → listam variáveis da sessão;
- `%xdel var` → apaga uma variável da memória;
- `%run script.py` → executa um script `.py` dentro do notebook;
- `%lsmagic` → lista todos os mágicos disponíveis.

```python
# O ? mostra a documentação (docstring) de qualquer objeto
# Teste também no console: brl?, brl??, print?, len?
brl?
```

```python
# %who e %whos: o que está na memória da sessão agora?
# %whos (verbose) mostra nome, tipo e conteúdo-resumo
%whos
```

```python
# %xdel remove uma variável da memória do kernel
temp = 27.216  # kg por bushel de soja — conversão usada no agro
print("temp criada:", temp)
%xdel temp
# Depois desta célula, "temp" não existe mais — rode %who na célula seguinte
# para conferir (se der NameError, é sinal de que o %xdel funcionou)
print("temp ainda existe?", "temp" in dir())
```

```python
# %lsmagic lista todos os comandos mágicos disponíveis
# A coluna "line" aceita %magic; a coluna "cell" exige %%magic (célula inteira)
%lsmagic
```

### 3.1 Experimento guiado: rodando um script com `%run`
O `%run` executa um script externo como se você tivesse colado o código na
célula. Para demonstrar, criamos um mini-script na pasta da semana com a função
`ipca_acumulado` que usaremos de novo na Aula 02:

```python
# %run executa um script .py e importa suas funções para a sessão
# (o %%writefile abaixo escreve o arquivo auxiliar na pasta da semana)
%%writefile util_aula01.py
def ipca_acumulado(taxas_mensais_pct):
    """Acumula taxas mensais (em %) — ex.: IPCA de 6 meses."""
    acumulado = 1.0
    for i in taxas_mensais_pct:
        acumulado *= (1 + i / 100)
    return (acumulado - 1) * 100

# teste rápido ao escrever o arquivo
if __name__ == "__main__":
    print(f"IPCA 6 meses (fev-jul/2026): {ipca_acumulado([0.70, 0.88, 0.67, 0.58, 0.16, 0.07]):.2f}%")
```

```python
# Roda o script e traz a função para dentro da sessão
%run util_aula01.py
# A função agora está disponível no namespace do notebook:
print(f"IPCA 12 meses via função: {ipca_acumulado([0.48, 0.09, 0.18, 0.33, 0.33, 0.70, 0.88, 0.67, 0.58, 0.16, 0.07]):.2f}%")
```

```python
# %lsmagic mostrou centenas de mágicos — os que usaremos no curso:
mágicos_do_curso = ["%who, %whos, %xdel  (inventário da sessão)",
                    "%run script.py                (executar arquivo .py)",
                    "%time instrucao               (tempo de UMA execução)",
                    "%timeit instrucao             (tempo médio de N execuções)",
                    "%%timeit                      (célula inteira cronometrada)",
                    "%lsmagic                      (lista tudo)"]
for m in mágicos_do_curso:
    print(m)
```

## 4. Cronometrando código: `%time` vs `%timeit`
Por que um economista se importaria com microssegundos? Porque em consultoria
as rotinas crescem: somar 1 milhão de lançamentos uma vez é instantâneo; somar
1 milhão de lançamentos **por dia, para 40 clientes**, já vira minutos que se
transformam em horas de processamento por mês. Medir antes de otimizar é o
primeiro passo de qualquer diagnóstico de performance.

Duas mágicas, dois usos (VanderPlas, Cap. 1; McKinney, Cap. 3, "Timing Code"):

- **`%time`** — roda a instrução **uma vez** e reporta o tempo de relógio
  (wall time). Bom para operações longas ou que não podem ser repetidas.
- **`%timeit`** — roda a instrução **várias vezes** e reporta a melhor média.
  Mais preciso; ideal para operações rápidas (microssegundos).

```python
# Preparação: 1 milhão de lançamentos contábeis fictícios (valores realistas)
lançamentos = [1200.50 + i * 0.01 for i in range(1_000_000)]  # R$ 1.200,50 + i centavos
print("tamanho da lista:", len(lançamentos))
print("primeiro:", lançamentos[0], "| último:", lançamentos[-1])
```

```python
# %time: UMA execução — bom para operações longas ou não repetíveis
%time total_py = sum(lançamentos)
print(f"Soma total: {brl(total_py)}")
```

```python
# %timeit: várias execuções — média mais confiável para operações rápidas
# (repare que o %timeit roda em silêncio e reporta N loops, best of M)
%timeit sum(lançamentos)
```

```python
# O mesmo cálculo com NumPy: converte a lista para array e soma vetorizado
arr = np.array(lançamentos)
%timeit np.sum(arr)   # espera-se ~10x mais rápido que sum() do Python puro
```

### 4.1 Leitura do resultado
O NumPy vence por uma ordem de grandeza ou mais: `np.sum` opera em um bloco
contíguo de memória com código compilado em C, enquanto `sum()` percorre uma
lista de objetos Python um a um. É o mesmo argumento do McKinney: em análise de
dados em larga escala, "aqueles milissegundos vão se acumulando".

**Regra prática do curso:** nunca otimize sem antes medir. `%timeit` para
microbenchmarks, `%time` para rodadas únicas. E lembre do conselho do Knuth
citado por VanderPlas: otimização prematura é a raiz de todo mal — primeiro
faça funcionar, depois cronometre, e só então otimize o gargalo real.

```python
# Forma de célula: %%timeit cronometra o bloco inteiro (deve ser a 1ª linha)
# Comparação final: loop manual vs sum() vs np.sum()
%%timeit
total_loop = 0
for x in lançamentos:
    total_loop += x
```

```python
# Veredito do experimento — rode as 3 medições e complete a tabela mental:
print("Loop manual em Python puro:  ~1 s por volta (lento)")
print("sum() embutida:              ~10 ms por volta (~100x mais rápida)")
print("np.sum(array):               ~1 ms por volta (~10x mais rápida que sum())")
print()
print("Conclusão: biblioteca vetorizada > função embutida > loop manual")
```

### Exercício 4 — Sua própria corrida de `%timeit`
Compare duas formas de somar os números de 1 a 100.000: (a) um loop `for` com
acumulador, cronometrado com `%%timeit` (célula inteira), e (b) a função
embutida `sum()`, cronometrada com `%timeit`. Qual é mais rápida e por quê?

```python
# EXERCÍCIO 4 — seu código aqui
```

```python
# SOLUÇÃO 4 — parte (a): o loop precisa de %%timeit (célula inteira)
# (%%timeit DEVE ser a primeira linha da célula)
numeros = range(1, 100_001)
%%timeit
total_loop = 0
for n in numeros:
    total_loop += n
```

```python
# SOLUÇÃO 4 — parte (b): sum() embutida com %timeit de linha
numeros = range(1, 100_001)
%timeit sum(numeros)
# Resultado esperado: sum() é ~10x+ mais rápida que o loop manual.
# Por quê: o for em Python puro paga o custo de interpretar cada passo;
# sum() é implementada em C e percorre a sequência uma única vez.
# Na Aula 07 (NumPy) a diferença sobe para ordens de grandeza.
```
